# Vehicle Fuel Consumption Tracker
# Problem Statement
# A transport company manages Car and Truck fleets. The fuel capacity of every vehicle should
# be private to prevent direct modification. An abstract base class will enforce a uniform method
# for computing fuel needed for trips.
# Your program should use:
# ● An abstract class Vehicle with an abstract method calculate_fuel(distance).
# ● Private instance variable __fuel_efficiency (km per liter) in Car.
# ● Getter and setter methods to safely update __fuel_efficiency.
# ● A concrete child class Car that implements calculate_fuel(distance) as
# distance / __fuel_efficiency.
# Outside the classes, the program should:
# 1. Create a Car object named "Toyota" with an efficiency of 15 km/l.
# 2. Calculate and print the fuel needed for a 150 km trip.
# Sample Output:
# Toyota Fuel Needed for 150 km: 10.00 Liters

```python
# Import the tools needed to create an abstract class
from abc import ABC, abstractmethod


# Parent abstract class
class Vehicle(ABC):

    # Abstract method
    # Every child class must implement this method
    @abstractmethod
    def calculate_fuel(self, distance):
        pass


# Child class that inherits from Vehicle
class Car(Vehicle):

    # Constructor
    def __init__(self, name, fuel_efficiency):
        self.name = name

        # Private instance variable
        self.__fuel_efficiency = fuel_efficiency

    # Getter method
    # Used to access the private fuel efficiency
    def get_fuel_efficiency(self):
        return self.__fuel_efficiency

    # Setter method
    # Used to safely change the private fuel efficiency
    def set_fuel_efficiency(self, fuel_efficiency):

        # Make sure the efficiency is greater than zero
        if fuel_efficiency > 0:
            self.__fuel_efficiency = fuel_efficiency
        else:
            print("Fuel efficiency must be greater than zero.")

    # Implement the abstract method from Vehicle
    def calculate_fuel(self, distance):

        # Fuel needed = distance / fuel efficiency
        return distance / self.__fuel_efficiency


# -----------------------------------------
# Outside the classes
# -----------------------------------------

# Create a Car object named Toyota
toyota = Car("Toyota", 15)

# Calculate fuel needed for a 150 km trip
fuel_needed = toyota.calculate_fuel(150)

# Print the result
print(f"{toyota.name} Fuel Needed for 150 km: {fuel_needed:.2f} Liters")
```
