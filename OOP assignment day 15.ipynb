```python

# Employee Salary Calculation using OOP
# Concepts Used:
# 1. Class
# 2. Inheritance
# 3. Class Variable
# 4. Method Overriding
# 5. Polymorphism


# Parent Class
class Employee:

    # Class variable (shared by all employees)
    tax_rate = 0.10

    # Constructor
    def __init__(self, name, basic_salary):
        self.name = name
        self.basic_salary = basic_salary

    # Method to calculate salary after tax
    def calculate_salary(self):
        final_salary = self.basic_salary - (self.basic_salary * Employee.tax_rate)
        return final_salary


# Child Class (inherits from Employee)
class SalesEmployee(Employee):

    # Constructor
    def __init__(self, name, basic_salary, commission):

        # Call the parent constructor
        super().__init__(name, basic_salary)

        # Instance variable unique to SalesEmployee
        self.commission = commission

    # Overriding the parent's calculate_salary() method
    def calculate_salary(self):

        total_salary = self.basic_salary + self.commission
        final_salary = total_salary - (total_salary * Employee.tax_rate)

        return final_salary



# Creating Objects


employee1 = Employee("John", 50000)

employee2 = SalesEmployee("Ada", 50000, 10000)


# Store objects in a list

employees = [employee1, employee2]



# Polymorphism
# Loop through each object and call
# calculate_salary()

for employee in employees:
 print(f"{employee.name}'s Final Salary is ₦{int(employee.calculate_salary())}")
#print(f"{employee.name}'s Final Salary is ₦{employee.calculate_salary}")
```
